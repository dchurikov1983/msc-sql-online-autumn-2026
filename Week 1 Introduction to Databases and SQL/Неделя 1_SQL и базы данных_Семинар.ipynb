```sql
DROP TABLE IF EXISTS {schema_name}.coffee_purchase;
DROP TABLE IF EXISTS {schema_name}.coffee;

CREATE TABLE {schema_name}.coffee (
    coffee_id INTEGER PRIMARY KEY,
    coffee_name TEXT NOT NULL,
    coffee_price [ТИП NUMERIC С ТОЧНОСТЬЮ 10 И МАСШТАБОМ 2] NOT NULL

CREATE TABLE {schema_name}.coffee_purchase (
    coffee_id INTEGER NOT NULL REFERENCES {schema_name}.coffee(coffee_id),
    order_time TIMESTAMP WITH TIME ZONE NOT NULL,
    vol [ЦЕЛОЧИСЛЕННЫЙ ТИП] NOT NULL,
    PRIMARY KEY (coffee_id, order_time)
);
